```python
fleet_size = int(input("Enter number of vehicles in fleet: "))
swaps_per_day = int(input("Enter swaps per vehicle per day: "))
avg_wait_time = float(input("Enter average wait time per swap (in minutes): "))
cost_per_minute = float(input("Enter cost per minute (₹): "))

baseline_wait = fleet_size * swaps_per_day * avg_wait_time
optimized_wait = baseline_wait * 0.7

time_saved = baseline_wait - optimized_wait
money_saved_per_day = time_saved * cost_per_minute
money_saved_per_month = money_saved_per_day * 30

print("Time saved per day:", time_saved)
print("Money saved per day:", money_saved_per_day)
print("Money saved per month:", money_saved_per_month)
```
