```python
import pandas as pd
import joblib
import numpy as np

# Load data
df = pd.read_csv("../data/simulated_data.csv")

# Load ML model
model = joblib.load("../wait_time_model.pkl")

# Features
X = df[["hour", "traffic_level", "rider_demand", "tariff_rs", "solar_kw"]]

# ML predictions
df["ml_predicted_wait"] = model.predict(X)

# Baseline logic: assume fixed wait (average)
baseline_wait = df["wait_time_min"].mean()
df["baseline_wait"] = baseline_wait

# Compare
df["wait_saved"] = df["baseline_wait"] - df["ml_predicted_wait"]

df[["hour", "baseline_wait", "ml_predicted_wait", "wait_saved"]]
```
